```python
import pandas as pd

file_path = r"D:\Swapnil\Work\Projects\FIFA\Updated Base Files\Coco-Cola.xlsx"

df = pd.read_excel(file_path)

print(df.head())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)
```

```python
df = df.rename(columns={
    "invoice_date": "date",
    "beverage_brand": "product",
    "region": "us_region"
})
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")

df["price_per_unit"] = pd.to_numeric(df["price_per_unit"], errors="coerce")
df["units_sold"] = pd.to_numeric(df["units_sold"], errors="coerce")
df["total_sales"] = pd.to_numeric(df["total_sales"], errors="coerce")
df["operating_profit"] = pd.to_numeric(df["operating_profit"], errors="coerce")
df["operating_margin"] = pd.to_numeric(df["operating_margin"], errors="coerce")
```

```python
df = df.dropna(subset=[
    "date",
    "product",
    "units_sold",
    "total_sales"
])
```

```python
df.insert(0, "sales_id", range(1, len(df) + 1))
```

```python
cols = [
    "sales_id",
    "date",
    "retailer",
    "retailer_id",
    "us_region",
    "state",
    "city",
    "product",
    "price_per_unit",
    "units_sold",
    "total_sales",
    "operating_profit",
    "operating_margin"
]

df = df[[c for c in cols if c in df.columns]]
```

```python
print(df.head())
print(df.shape)
print(df.dtypes)
print(df.isna().sum())
```

```python
df.to_excel("step1_sales_base.xlsx", index=False)
```
